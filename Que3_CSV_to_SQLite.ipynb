```python
import csv
import re
import sqlite3
```

```python
conn = sqlite3.connect("users.db")
cursor = conn.cursor()

cursor.execute("""
    CREATE TABLE IF NOT EXISTS users (
        id    INTEGER PRIMARY KEY AUTOINCREMENT,
        name  TEXT NOT NULL,
        email TEXT NOT NULL UNIQUE
    )
""")
conn.commit()
print("Table ready")
```

```python
EMAIL_PATTERN = re.compile(r"^[^@\s]+@[^@\s]+\.[^@\s]+$")
inserted = duplicates = invalid = 0

with open("users.csv", newline="", encoding="utf-8-sig") as f:
    reader = csv.DictReader(f)
    for row in reader:
        name = (row.get("name") or "").strip()
        email = (row.get("email") or "").strip().lower()

        # Skip rows with missing name or bad email
        if not name or not EMAIL_PATTERN.match(email):
            print(f"Skipping invalid row: {row}")
            invalid += 1
            continue

        # The ? placeholders protect against SQL injection
        cursor.execute(
            "INSERT OR IGNORE INTO users (name, email) VALUES (?, ?)",
            (name, email),
        )
        if cursor.rowcount == 1:
            inserted += 1
        else:
            duplicates += 1

conn.commit()
print(f"Inserted: {inserted} | Duplicates skipped: {duplicates} | Invalid skipped: {invalid}")
```
